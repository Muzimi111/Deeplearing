### 1. np.unravel() 函数的使用

np.unravel_index(indices, shape, order = ‘C’)
一句话概括：求出数组某元素（或某组元素）拉成一维后的索引值在原本维度（或指定新维度）中对应的索引
- indices: 整数构成的数组， 其中元素是索引值（integer array whose elements are indices into flattened version of array)
- shape: tuple of ints, 一般是原本数组的维度，也可以给定的新维度。


```python
import numpy as np

np.random.seed(0)  # 设置随机种子以确保结果可重复
A = np.random.randint(1, 100, size = (2, 3, 5)) # randint是生成随机整数的函数，1-100之间，size是生成的数组形状
print(A)
print(f"最大值: {np.max(A)}")  # 输出最大值
# argmax函数返回的是最大值的索引，而不是最大值本身。
# 对于多维数组，argmax会将数组展平为一维，然后返回展平后数组中最大值的索引。如果你想在特定的维度上找到最大值的索引，可以使用axis参数。
ind_max = np.argmax(A) # 在第3维度上求最大值的索引
print(f"最大值的索引: {ind_max}")
# 使用 np.unravel_index 将展平后的索引转换为多维索引
multi_dim_index = np.unravel_index(ind_max, A.shape)
print(f"最大值的多维索引: {multi_dim_index}")  # 输出 (0, 2, 4) 表示在第0个二维数组中，第2行，第4列的元素是最大值。
print(f"最大值对应的元素: {A[multi_dim_index]}")  # 输出最大值对应的元素
```

index可以升维,假设取indices = (4, 7), 此时np.unravel_index输出的结果为3个数组（3即为A的维度），每个数组的第0位为下标为4的元素在A.shape下的下标， 每个数组的第1位为下标为7的元素在A.shape下的下标

```python
inds = np.unravel_index((4, 7), A.shape) # 将展平后的索引 (4, 7) 转换为多维索引
print(inds)  # 输出 (1, 1, 2) 表示在第1个二维数组中，第1行，第2列的元素对应展平后的索引 4 和 7。
first_index = (inds[0][0], inds[1][0], inds[2][0])  # 第一个维度的索引
second_index = (inds[0][1], inds[1][1], inds[2][1])  # 第二个维度的索引
print(f"第一个索引对应的元素: {A[first_index]}")  # 输出第一个索引对应的元素
print(f"第二个索引对应的元素: {A[second_index]}")  # 输出第二个索引对应的元素
```

### 2. np.arange() 和 np.linspace() 的区别

```python
# np.arange() 函数用于创建一个包含等间隔值的数组。它的基本语法是：
# np.arange([start, ]stop, [step, ]dtype=None)
# 参数说明：
# start: 可选参数，表示起始值，默认为 0。
# stop: 必需参数，表示终止值（不包含该值）。
# step: 可选参数，表示步长，默认为 1。
# dtype: 可选参数，表示输出数组的数据类型。
arr1 = np.arange(0, 10, 2)  # 创建一个从 0 到 10（不包含 10），步长为 2 的数组
print(arr1)  # 输出: [0 2 4 6 8]
```

```python
# np.linspace() 函数用于创建一个包含指定数量的等间隔值的数组。它的基本语法是：
# np.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)
# 参数说明：
# start: 必需参数，表示起始值。
# stop: 必需参数，表示终止值。
# num: 可选参数，表示要生成的等间隔样本数量，默认为是 50。
# endpoint: 可选参数，布尔值，表示是否包含终止值，默认为 True。
# retstep: 可选参数，布尔值，表示是否返回步长，默认为 False。
# dtype: 可选参数，表示输出数组的数据类型。
# axis: 可选参数，表示沿着哪个轴生成样本，默认为 0。
arr2 = np.linspace(0, 1, num=4)  # 创建一个从 0 到 1（包含 1），等间隔分成 5 个样本的数组
print(arr2)  # 输出: [0.   0.25 0.5  0.75 1.  ] 
```

总结：当强调步长时使用arange，强调开始和结束时使用linspace。后者默认包含结束，前者左闭右开。

### 3.tqdm的使用

```python
import tqdm

# 使用方法
# 1. 迭代一个列表
for i in tqdm.tqdm([1, 2, 3, 4, 5]):
    # 模拟一些工作
    pass    
for i in tqdm.tqdm(range(100)):
    # 模拟一些工作
    pass
# trange() 是 tqdm 库提供的一个函数，它是对 Python 内置的 range() 函数的一个增强版本。trange() 的作用是创建一个带有进度条的迭代器，方便在循环中显示进度信息。它的基本用法如下：
from tqdm import trange
for i in trange(100):
    # 模拟一些工作
    pass    
```

```python
# 2. 为进度条设置描述
for i in tqdm.tqdm(range(100), desc="Processing"):
    # 模拟一些工作
    pass

import time
pbar = tqdm.tqdm(["a", "b", "c", "d", "e"])
for item in pbar:
    pbar.set_description(f"Processing {item}")
    time.sleep(0.5)  # 模拟一些工作
```

```python
# 3. 手动控制进度
# pbar是一个tqdm对象，可以通过update()方法手动更新进度条的进度。下面是一个示例：
# tqdm.tqdm(total=100) 创建了一个总进度为 100 的进度条对象。然后在循环中，每次迭代都会调用 pbar.update(10) 来更新进度条的进度，表示已经完成了 10 个单位的工作。最后，使用 pbar.close() 来关闭进度条。
with tqdm.tqdm(total=100) as pbar:
    for i in range(10):
        # 模拟一些工作
        time.sleep(0.1)
        pbar.update(10)  # 每次更新进度条 10 个单位
# update() 方法可以接受一个整数参数，表示要增加的进度量。每次调用 update(n) 时，进度条会向前移动 n 个单位。
```

```python
# 4， write() 方法用于在进度条上方输出信息，而不会干扰进度条的显示。它的基本用法如下：
with tqdm.tqdm(total=100) as pbar:
    for i in range(10):
        # 模拟一些工作
        time.sleep(0.1)
        pbar.update(10)  # 每次更新进度条 10 个单位
        pbar.write(f"Completed {i + 1} iterations")  # 在进度条上方输出信息

# 除了使用with语句来创建进度条对象外，还可以直接创建一个tqdm对象，然后在循环中手动更新进度条。下面是一个示例：
pbar = tqdm.tqdm(total=100)
for i in range(10):
    # 模拟一些工作
    time.sleep(0.1)
    pbar.update(10)  # 每次更新进度条 10 个单位
    pbar.write(f"Completed {i + 1} iterations")  # 在进度条上方输出信息
pbar.close()  # 关闭进度条
```

```python
# 6. 自定义进度条显示信息
# 通过set_description()方法可以自定义进度条的描述信息。下面是一个示例：
# set_postfix() 方法用于在进度条的右侧显示附加信息。它的基本用法如下：
pbar = tqdm.tqdm(total=100)
for i in range(10):
    # 模拟一些工作
    pbar.update(10)  # 每次更新进度条 10 个单位
    pbar.set_description(f"Processing {i + 1}")  # 设置进度条的描述信息
    pbar.set_postfix({"Iteration": i + 1, "Status": "Running"})  # 设置附加信息，postfix可以接受一个字典，字典的键值对会显示在进度条的右侧。
pbar.close()  # 关闭进度条
```
