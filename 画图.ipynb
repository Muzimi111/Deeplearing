### 画图
matplotlib架构：从下往上分别是Backend，Artist和Scripting。
- Backend Layer有三个内建的抽象类别：
  - FigureCanvas： 负责管理画布的大小，背景色等属性
  - Renderer： 负责将FigureCanvas的艺术家对象渲染成实际图形
  - Event： 处理使用者输入
- Artist Layer分为两种主要形态
  - 基本艺术家
  - 组合艺术家
  


- Figure是最顶层的整体视窗，可以包含多个Axes,Axes是绘制数据与相关刻度，标签的区域，可以将其视为一个坐标系统；Axis是坐标系统中的坐标轴

```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure( facecolor='w', edgecolor='k')
print(fig)
plt.show()
```

```python
fig = plt.figure(figsize=(8, 6), facecolor='y', edgecolor='r')
ax = fig.add_subplot(111)
ax.set(xlabel='X-axis', ylabel='Y-axis', title='Sample Plot', xlim=(0, 10), ylim=(0, 10))
plt.show()
```

```python
# Figure建立方式
# 1. 显性方式
fig = plt.figure()
# 2. 隐性方式
plt.subplot()
```

```python
# 3. 使用plt.axes()创建子图
ax1 = plt.axes()
ax2 = plt.axes([0.5, 0.5, 0.4, 0.4], facecolor='y')  # [left, bottom, width, height]
plt.show()
```

```python
# fig.add_axes()方法创建子图
fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])
plt.show()
```

```python
# plt.subplots()创建子图
# 与subplot()不同，subplots()返回一个Figure对象和一个Axes对象数组，不是一次只创建一个子图，而是一次创建多个子图，并用数组的形式返回
# 其方法是通过设定nrows和ncols参数来指定子图的行数和列数，返回的Axes对象数组的形状与nrows和ncols相同
# 另外也可以使用sharex和sharey参数来指定子图之间是否共享x轴和y轴
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(8, 6), sharex=True, sharey=True)
ax[0, 0].set(title='Subplot 1', xlabel='X-axis', ylabel='Y-axis')
ax[0, 1].set(title='Subplot 2', xlabel='X-axis', ylabel='Y-axis')
ax[1, 0].set(title='Subplot 3', xlabel='X-axis', ylabel='Y-axis')
ax[1, 1].set(title='Subplot 4', xlabel='X-axis', ylabel='Y-axis')
plt.tight_layout()  
plt.show()
```

```python
fig_l = plt.figure(figsize=(8, 6))
x = np.linspace(0, 10, 100)
h = 3
colors = ['b', 'g', 'r', 'c', 'm', 'y', 'k']
for i, k in enumerate(np.linspace(4, 10, 7)):
    p1 = 1/(1 + np.exp(-k*(x - h)))
    plt.plot(x, p1, label=f'k={k:.2f}', color=colors[i])
    t = k - 4
    p2 = 1/(1 + np.exp(-t*(x - h))) + t
    plt.plot(x, p2, label=f't={t:.2f}', color=colors[i], linestyle='--')
plt.legend()
plt.show()
```
