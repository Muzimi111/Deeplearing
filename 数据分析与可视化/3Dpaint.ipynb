### 创建三维坐标轴对象Axes3D
1. 利用关键字`projection = '3d'`来实现
2. 从`mpl_toolkits.mplot3d`导入对象`Axes3D`来实现

```python
#方法一，利用关键字导入相关模块绘制
from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 定义坐标轴
fig = plt.figure()
ax1 = plt.axes(projection='3d')
# 画多个子图
# ax = fig.add_subplot(111, projection='3d')
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure()
ax1 = plt.axes(projection='3d')

z = np.linspace(0,13,1000)
x = 5*np.sin(z)
y = 5*np.cos(z)

zd = 13*np.random.random(100)
xd = 5*np.sin(zd)
yd = 5*np.cos(zd)

ax1.scatter3D(xd,yd,zd, c = 'r')  #绘制散点图
ax1.plot3D(x,y,z,'gray')    #绘制空间曲线
plt.show()
```

```python
fig = plt.figure()  #定义新的三维坐标轴
ax3 = plt.axes(projection='3d')

xx = np.arange(-5, 5, 0.25)
yy = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(xx, yy)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

ax3.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap='rainbow')  
# 设置坐标轴范围
ax3.set_xlim(-5, 5)
ax3.set_ylim(-5, 5)
ax3.set_zlim(-2, 2)
# 设置坐标轴标签
ax3.set_xlabel('X')
ax3.set_ylabel('Y')

#绘制三维曲面,rstride和cstride分别表示行和列的跨度，越小绘图越精细，cmap表示颜色映射
# contour是绘制等高线的函数，参数分别为X,Y,Z,10表示等高线的数量，colors表示等高线的颜色，linestyles表示等高线的样式，offset表示等高线在Z轴上的偏移量
ax3.contour(X, Y, Z, 10, colors='black', linestyles='solid', offset=-2)
plt.show()
```
